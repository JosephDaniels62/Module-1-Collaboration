# Pandas for Data Science in Python

This notebook follows along with a beginner Pandas tutorial using the provided `telco_churn.csv` dataset.

Topics covered:
- Importing pandas
- Loading a CSV file
- Viewing rows and columns
- Checking data types
- Descriptive statistics
- Selecting columns
- Filtering data
- Grouping and aggregation
- Basic insights


```
import pandas as pd

# Load dataset
df = pd.read_csv("telco_churn.csv")

# Display first 5 rows
df.head()
```

```
# Shape of dataset (rows, columns)
df.shape
```

```
# Column names
df.columns
```

```
# Data types and non-null counts
df.info()
```

```
# Summary statistics for numeric columns
df.describe()
```

```
# Check missing values
df.isnull().sum()
```

```
# Select specific columns
df[['customerID', 'gender', 'tenure', 'MonthlyCharges']].head()
```

```
# Filter customers with MonthlyCharges greater than 80
high_bill = df[df['MonthlyCharges'] > 80]

high_bill.head()
```

```
# Count churned vs non-churned customers
df['Churn'].value_counts()
```

```
# Average monthly charges grouped by churn status
df.groupby('Churn')['MonthlyCharges'].mean()
```

```
# Group by contract type and check average monthly charges
df.groupby('Contract')['MonthlyCharges'].mean()
```

## Quick Observations

- We loaded the Telco Churn dataset using pandas.
- We explored the structure using `.head()`, `.shape()`, `.columns()`, and `.info()`.
- We checked summary statistics with `.describe()`.
- We filtered rows based on billing values.
- We used `groupby()` to compare churn and contract trends.

This demonstrates the core beginner workflow for using Pandas in data science.
