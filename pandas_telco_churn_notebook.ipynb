# Pandas for Data Science in Python

In this notebook, I followed along with a beginner Pandas tutorial using the provided `telco_churn.csv` dataset.

I used this project to practice the basic Pandas skills that are commonly used in data science, including:

- Importing pandas
- Loading a CSV file
- Viewing rows and columns
- Checking data types
- Looking at descriptive statistics
- Selecting specific columns
- Filtering data
- Grouping and aggregation
- Finding basic insights from the dataset

This helped me get more comfortable working with datasets inside Jupyter Notebook.


```
import pandas as pd

# Load dataset
df = pd.read_csv("telco_churn.csv")

# Display first 5 rows
df.head()
```

```
# Shape of dataset (rows, columns)
df.shape
```

```
# Column names
df.columns
```

```
# Data types and non-null counts
df.info()
```

```
# Summary statistics for numeric columns
df.describe()
```

```
# Check missing values
df.isnull().sum()
```

```
# Select specific columns
df[['customerID', 'gender', 'tenure', 'MonthlyCharges']].head()
```

```
# Filter customers with MonthlyCharges greater than 80
high_bill = df[df['MonthlyCharges'] > 80]

high_bill.head()
```

```
# Count churned vs non-churned customers
df['Churn'].value_counts()
```

```
# Average monthly charges grouped by churn status
df.groupby('Churn')['MonthlyCharges'].mean()
```

```
# Group by contract type and check average monthly charges
df.groupby('Contract')['MonthlyCharges'].mean()
```
